```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from google.colab import files
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

np.random.seed(42)
tf.random.set_seed(42)

# Upload your downloaded Kaggle dataset
uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Clean column names
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Available columns:", df.columns.tolist())

# Find the date column
date_candidates = [
    "date", "datetime", "timestamp", "trade_date"
]

date_col = next(
    (c for c in date_candidates if c in df.columns),
    None
)

# Find the closing-price column
close_candidates = [
    "close", "closing_price", "close_price",
    "adj_close", "adjusted_close", "adjclose"
]

close_col = next(
    (c for c in close_candidates if c in df.columns),
    None
)

if close_col is None:
    raise ValueError(
        "Closing-price column not found. "
        "Check the printed column names and set close_col manually."
    )

if date_col is None:
    raise ValueError(
        "Date column not found. Set date_col to your dataset's date column."
    )

# Convert date and closing price into proper formats
df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
df[close_col] = pd.to_numeric(df[close_col], errors="coerce")

# Optional: select one stock if the CSV contains multiple tickers
ticker_col = next(
    (c for c in ["symbol", "ticker", "stock_symbol"]
     if c in df.columns),
    None
)

STOCK_SYMBOL = None  # Example: "AAPL"; None selects the first available ticker

if ticker_col is not None:
    symbols = df[ticker_col].dropna().unique()
    print("Available stock symbols:", symbols)

    if STOCK_SYMBOL is None and len(symbols) > 0:
        STOCK_SYMBOL = symbols[0]

    if STOCK_SYMBOL is not None:
        df = df[
            df[ticker_col].astype(str).str.upper()
            == STOCK_SYMBOL.upper()
        ].copy()

        print("Selected stock:", STOCK_SYMBOL)

# Remove missing values and duplicate dates
df = df.dropna(subset=[date_col, close_col])
df = df.drop_duplicates(subset=[date_col], keep="last")

# Remove invalid closing prices
df = df[df[close_col] > 0]

# Sort chronologically
df = df.sort_values(date_col).reset_index(drop=True)

# Select 2,000 consecutive available trading records
if len(df) < 2000:
    raise ValueError(
        f"Only {len(df)} valid records found. "
        "At least 2,000 are required."
    )

df = df.tail(2000).copy()
df = df.reset_index(drop=True)

print("\nCleaned dataset shape:", df.shape)
print("First date:", df[date_col].min())
print("Last date:", df[date_col].max())
print("Number of records:", len(df))

display(df[[date_col, close_col]].head())
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    df[date_col],
    df[close_col],
    label="Historical Closing Price"
)

plt.title("Stock Closing Price - 2000 Records")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
prices = df[[close_col]].values.astype("float32")
dates = df[date_col].values

LOOKBACK = 60

# Chronological 80% training / 20% testing
split_index = int(len(prices) * 0.80)

scaler = MinMaxScaler(feature_range=(0, 1))

# Fit using training data only
scaler.fit(prices[:split_index])

# Transform all prices using the training-fitted scaler
scaled_prices = scaler.transform(prices)

print("Training records:", split_index)
print("Testing records:", len(prices) - split_index)
print("Window size:", LOOKBACK, "trading days")
print("Normalized range:", scaled_prices.min(), "to", scaled_prices.max())
```

```python
X = []
y = []
target_indices = []

for i in range(LOOKBACK, len(scaled_prices)):
    X.append(scaled_prices[i - LOOKBACK:i])
    y.append(scaled_prices[i, 0])
    target_indices.append(i)

X = np.array(X, dtype=np.float32)
y = np.array(y, dtype=np.float32)
target_indices = np.array(target_indices)

# Keep the split chronological
train_mask = target_indices < split_index
test_mask = target_indices >= split_index

X_train = X[train_mask]
y_train = y[train_mask]

X_test = X[test_mask]
y_test = y[test_mask]

test_indices = target_indices[test_mask]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
model = Sequential([
    LSTM(
        64,
        return_sequences=True,
        input_shape=(LOOKBACK, 1)
    ),
    Dropout(0.2),

    LSTM(32),
    Dropout(0.2),

    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)

model.summary()

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_split=0.1,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)

print("Training completed!")
```

```python
plt.figure(figsize=(10, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("LSTM Training Performance")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Predict normalized prices
predicted_scaled = model.predict(X_test, verbose=0)

# Convert normalized prices back to original price units
predicted_prices = scaler.inverse_transform(predicted_scaled)

actual_prices = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)

# Evaluation metrics
mae = mean_absolute_error(
    actual_prices.flatten(),
    predicted_prices.flatten()
)

rmse = np.sqrt(
    mean_squared_error(
        actual_prices.flatten(),
        predicted_prices.flatten()
    )
)

print("MODEL EVALUATION")
print("-" * 35)
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print("-" * 35)

print("MAE = Mean Absolute Error")
print("RMSE = Root Mean Squared Error")
```

```python
test_dates = pd.to_datetime(dates[test_indices])

plt.figure(figsize=(15, 6))

plt.plot(
    test_dates,
    actual_prices.flatten(),
    label="Actual Closing Price",
    linewidth=2
)

plt.plot(
    test_dates,
    predicted_prices.flatten(),
    label="LSTM Predicted Price",
    linewidth=2
)

plt.title("Actual vs Predicted Stock Closing Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Use the most recent 60 observed closing prices
last_window = scaled_prices[-LOOKBACK:].copy()

future_predictions = []

# Recursive forecasting: each prediction becomes the next input
for day in range(10):
    input_data = last_window.reshape(1, LOOKBACK, 1)

    next_price_scaled = model.predict(
        input_data,
        verbose=0
    )[0, 0]

    future_predictions.append(next_price_scaled)

    # Slide the window forward by one day
    last_window = np.vstack([
        last_window[1:],
        [[next_price_scaled]]
    ])

# Convert forecasts to original price units
future_predictions = scaler.inverse_transform(
    np.array(future_predictions).reshape(-1, 1)
).flatten()

# Generate future weekdays as estimated trading dates
last_date = pd.Timestamp(df[date_col].iloc[-1])

future_dates = pd.bdate_range(
    start=last_date + pd.Timedelta(days=1),
    periods=10
)

forecast_df = pd.DataFrame({
    "Future Date": future_dates,
    "Predicted Closing Price": future_predictions
})

print("NEXT 10 TRADING-DAY FORECAST")
display(forecast_df.round(2))
```

```python
plt.figure(figsize=(15, 6))

# Last 100 observed prices
plt.plot(
    df[date_col].iloc[-100:],
    df[close_col].iloc[-100:],
    label="Historical Closing Price"
)

# Forecast for the next 10 business days
plt.plot(
    future_dates,
    future_predictions,
    marker="o",
    linestyle="--",
    label="Next 10 Days Forecast"
)

plt.axvline(
    x=last_date,
    linestyle=":",
    label="Forecast Starts"
)

plt.title("Stock Price Forecast Using LSTM")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Save test predictions
test_results = pd.DataFrame({
    "Date": test_dates,
    "Actual Closing Price": actual_prices.flatten(),
    "Predicted Closing Price": predicted_prices.flatten()
})

test_results.to_csv(
    "test_predictions.csv",
    index=False
)

# Save future forecast
forecast_df.to_csv(
    "future_10_day_forecast.csv",
    index=False
)

# Save the trained model
model.save("stock_lstm_model.keras")

print("Files saved successfully!")

# Download the forecast and evaluation results
files.download("test_predictions.csv")
files.download("future_10_day_forecast.csv")
files.download("stock_lstm_model.keras")
```
